```python
contacts = {
    "Arjun": "1234",
    "Sumit": "555",
    "Abhay": "83993",
    "Adity": "8138833829"
}
search_name = input("Enter a name to search: ").strip()
found = False
for name, phone in contacts.items():
    if name.lower() == search_name.lower():
        print(f"Phone number for {name}: {phone}")
        found = True
        break

if not found:
    print(f"Sorry, no contact found for '{search_name}'.")
```

```python
def my_abs(val):
    if val < 0:
        return -val
    return val

print(my_abs(10))
print(my_abs(-7))
print(my_abs(0))
```
