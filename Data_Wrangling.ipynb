# Assignment 5 - Data Wrangling



```python
import numpy as np
import pandas as pd
from pandas import DataFrame, Series
```

# Part 1

# Q1

```python
students_master=pd.read_csv("students_master.csv")
```

```python
exam_marks=pd.read_csv("exam_marks.csv")
```

```python
display(students_master.head())
display(exam_marks.head())
```

```python
common_column=list(set(students_master.columns)&set(exam_marks.columns))
display(common_column)
```

# Q2

```python
merged_inner = pd.merge(students_master, exam_marks, on="Student_ID", how="inner")
display(merged_inner)
```

# Q3

```python
merged_left = pd.merge(students_master, exam_marks, on="Student_ID", how="left")
display(merged_left)
```

# Q4

```python
merged_right = pd.merge(students_master, exam_marks, on="Student_ID", how="right")
display(merged_right)
```

# Q5

```python
merged_outer = pd.merge(students_master, exam_marks, on="Student_ID", how="outer")
display(merged_outer["Student_ID"].unique())
```

# Q6

```python
section_CA = pd.read_csv("section_CA.csv")
section_CB = pd.read_csv("section_CB.csv")
concat_v_orig = pd.concat([section_CA, section_CB], axis=0)
display(concat_v_orig)
```

# Q7

```python
concat_v_ignore = pd.concat([section_CA, section_CB], axis=0, ignore_index=True)
display(concat_v_ignore.index)
display(concat_v_ignore)
```

# Q8

```python
concat_h = pd.concat([section_CA, section_CB], axis=1)
display(concat_h.columns)
display(concat_h)
```

# Q9

```python
s1 = section_CA["Attendance"]
s2 = section_CB["Student_ID"]
series_concat = pd.concat([s1, s2], axis=0)
display(series_concat)
```
