# Section B

# Q16

```python
branch_units_stats = df.groupby("branch")["units_sold"].agg(["sum", "mean", "min", "max", "std"])
display(branch_units_stats)
```

# Q17

```python
product_multi_agg = df.groupby("product").agg(
    total_units_sold=("units_sold", "sum"),
    avg_discount_pct=("discount_pct", "mean"),
    max_customer_rating=("customer_rating", "max"),
    min_delivery_time=("delivery_minutes", "min")
)
display(product_multi_agg)
```

# Q18

```python
branch_size = df.groupby("branch").size()
branch_count = df.groupby("branch")["customer_rating"].count()
size_vs_count = pd.DataFrame({
    "total_records": branch_size,
    "non_missing_ratings": branch_count,
    "missing_ratings": branch_size - branch_count
})
display(size_vs_count)
```

# Q19

```python
def calc_range(s):
    return s.max() - s.min()

product_units_range = df.groupby("product")["units_sold"].agg(calc_range)
display(product_units_range)
```
